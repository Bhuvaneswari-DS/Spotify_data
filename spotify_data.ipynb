```python
correlation_matrix = dfm.corr(numeric_only=True)

# Display heatmap (optional, with seaborn)
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation Matrix of Numeric Features")
plt.show()
```

```python
# Additional Seaborn visualizations for deeper insights

fig, axes = plt.subplots(3, 1, figsize=(14, 20))


# 2. Count of Songs by Emotion
sns.countplot(data=dfm, x='emotion', order=dfm['emotion'].value_counts().index, ax=axes[1], palette='Set2')
axes[1].set_title("Count of Songs by Emotion")
axes[1].tick_params(axis='x', rotation=45)
```

```python
import plotly.express as px
```

```python
emotion_by_year = dfm.groupby(['Release Date','Danceability','emotion','Positiveness']).size().reset_index(name='count')
```

```python
fig = px.bar(
    emotion_by_year,
    x='Danceability',
    y='count',
    color='emotion',
    animation_frame='Release Date',
    range_y=[0, emotion_by_year['count'].max()],
    title='Emotion Distribution Over Years'
)
fig.show()
```
